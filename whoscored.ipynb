```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time

driver = webdriver.Chrome()
driver.get("https://www.whoscored.com/Matches")

time.sleep(5)  # attendre le chargement JS

matches = driver.find_elements(By.CLASS_NAME, "match-link")

for m in matches:
    print(m.text)

driver.quit()
```
