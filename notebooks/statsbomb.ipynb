```python
import json
import pandas as pd
import os
```

```python
lineups_path = "statsbomb_datasets/data/lineups"

players_list = []

for file in os.listdir(lineups_path):
    if file.endswith(".json"):
        with open(os.path.join(lineups_path, file), encoding="utf-8") as f:
            data = json.load(f)

            for team in data:
                team_name = team["team_name"]

                for player in team["lineup"]:
                    players_list.append({
                        "player_id": player.get("player_id"),
                        "player_name": player.get("player_name"),
                        "team": team_name,
                        "position": player.get("position", {}).get("name"),
                        "jersey_number": player.get("jersey_number")
                    })

# On crée le dataframe des joueurs
df_players = pd.DataFrame(players_list)

# On supprime les doublons
df_players = df_players.drop_duplicates(subset=["player_id"])

df_players.head()
```

```python
# Les events sont les matchs donc toutes les données de performance
# sont des les jsons de events

events_path = "statsbomb_datasets/data/events"

all_dfs = []

for file in os.listdir(events_path):
    # On prend que les fichiers json
    if not file.endswith(".json"):
        continue

    path = os.path.join(events_path, file)

    with open(path, encoding="utf-8") as f:
        data = json.load(f)
    
    # J'essaye de récupérer les données de performance des joueurs
    df = pd.json_normalize(data)
    all_dfs.append(df)

# On concatène les données
df_all = pd.concat(all_dfs, ignore_index=True)

# J'essaye de créer un parquet (sorte de csv)
df_all.to_parquet("statsbomb_events.parquet", index=False)
```
