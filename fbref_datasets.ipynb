```python
import pandas as pd
from pathlib import Path
```

```python
folder = Path("fbref_datasets")
summary = []

# On regarde le nombre de lignes, colonnes et NA apr fichier csv

for file in folder.glob("*.csv"):
    df = pd.read_csv(file)

    summary.append({
        "fichier": file.name,
        "lignes": len(df),
        "colonnes": len(df.columns),
        "NA": df.isna().sum().sum()
    })

print(pd.DataFrame(summary))
```
