```python
import pandas as pd

df = pd.read_csv("male_players.csv")

# Les 5 premières lignes du dataset
print("5 premières lignes")
print(df.head())

```

```python
#Nombre de lignes et colonnes
print(f"\n Dimensions")
print(f"Lignes : {len(df)}")
print(f"Colonnes : {len(df.columns)}")

# Liste des colonnes
print(f"\nColonnes ({len(df.columns)})")
print(df.columns.tolist())

# Détection des NaN
print("\nValeurs manquantes (NaN) par colonne")
nan_counts = df.isnull().sum()
print(nan_counts[nan_counts > 0])  # Affiche uniquement les colonnes avec des NaN

print(f"\nTotal NaN dans le dataset : {df.isnull().sum().sum()}")
print(f"Pourcentage de valeurs manquantes : {df.isnull().mean().mean() * 100:.2f}%")
```
